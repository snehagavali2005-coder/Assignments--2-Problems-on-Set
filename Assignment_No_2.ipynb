# SECTION C -ADVANCED

```python
a = {100, 4, 200, 1, 3, 2}
longest = []

for num in a:
    if num - 1 not in a:
        current = num
        sequence = []

        while current in a:
            sequence.append(current)
            current += 1

        if len(sequence) > len(longest):
            longest = sequence

print("Sequence =", longest)
print("Length =", len(longest))
```

```python
numbers = [2, 7, 11, 15, 1, 8]
target = 9

seen = set()
pairs = set()

for num in numbers:
    diff = target - num

    if diff in seen:
        pairs.add(tuple(sorted((num, diff))))

    seen.add(num)

print("Pairs =", pairs)
```

```python
a = {1, 2, 3}
power_set = [set()]

for num in a:
    for subset in power_set.copy():
        power_set.append(subset | {num})

print("Number of subsets =", len(power_set))
print("Subsets include:")

for subset in power_set:
    print(subset)
```

```python
a = [1, 2, 1, 3, 4, 2, 3]
k = 4
result = []

for i in range(len(a) - k + 1):
    window = a[i:i+k]
    result.append(len(set(window)))

print("Distinct counts =", result)
```

```python
a = [1, 2, 2, 4, 5]
n = 5

original = set(a)
expected = set(range(1, n + 1))

duplicate = sum(a) - sum(original)
missing = expected - original

print("Duplicate =", duplicate)
print("Missing =", missing.pop())
```

```python
universe = {1, 2, 3, 4, 5}
groups = [{1, 2}, {3, 4}, {5}]

combined = set()
valid = True

for group in groups:
    if combined.intersection(group):
        valid = False
    combined.update(group)

if combined != universe:
    valid = False

print("Valid disjoint partition =", valid)
```

```python
universe = {1, 2, 3, 4, 5}

S1 = {1, 2, 3}
S2 = {2, 4}
S3 = {3, 4, 5}

sets = {'S1': S1, 'S2': S2, 'S3': S3}
names = list(sets.keys())

for i in range(len(names)):
    for j in range(i + 1, len(names)):
        a = sets[names[i]]
        b = sets[names[j]]

        if a.union(b) == universe:
            print("Smallest covering pair =", (names[i], names[j]))
            break
    else:
        continue
    break
```
